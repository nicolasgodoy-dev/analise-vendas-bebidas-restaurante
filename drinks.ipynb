```python
#Importar pandas
import pandas as pd
```

```python
#Ler arquivo "dados.xlsx" e a planilha "Por Canal"
df1 = pd.read_excel('dados.xlsx', sheet_name='Por Canal')
df1
```

```python
#Ler arquivo "dados.xlsx" e a planilha "Consumo de Produtos"
df2 = pd.read_excel('dados.xlsx', sheet_name='Consumo de Produtos')
df2
```

```python
#filtrar categorias desejadas
df2 = df2[df2['T. Categoria'].isin(['DRINKS', 'SOFT DRINKS'])]
df2
```

```python
# Calcular total de clientes atendidos por loja
n_pessoas = df1.groupby('Estabelecimento')['N.Pessoas'].sum()
n_pessoas
```

```python
# Calcular quantidade de produtos vendidos por loja e categoria
quant = df2.groupby(['Estabelecimento','T. Categoria'])['Quant.'].sum()
quant
```

```python
#Quantidade de pedidos por cliente
quant_por_cliente = quant / n_pessoas
quant_por_cliente
```

```python
#Plotar grafico para visualizar os resultados

ax = quant_por_cliente.plot.barh() 
#ax = axes
#Series.plot.bar() assim ja podendo exibir
#bar() exibe com os titulos na vertical enquanto barch() exibe na horizontal

#bar_label adiciona rotulos
#ax.containers[0] é o elemento
#fmt='{:.2f}' é o formato que vai sair as informações sendo apenas 2 numeros após o ponto
ax.bar_label(ax.containers[0], fmt='{:.2f}')

#Aumenta o grafico em 10%
#Primeiro argumento sendo posição inicial e o segundo posição final.
ax.set_xlim(0, ax.get_xlim()[1] * 1.1);# ";" para não aparecer o resultado do set_xlim()


```

```python
#Mostrar top 5 Produtos por loja e categoria

#Agrupa os valores e soma os valores de 'Quant.
top5 = df2.groupby(['Estabelecimento','T. Categoria', 'Descrição'])['Quant.'].sum()

#sort é para ordenar os valores, por padrão ele ordendena em ascedente.
top5 = top5.sort_values(ascending=False)#false para que seja do maior para o menor

#Agrupar por estabelecimento e categoria mostrando os 5 primeiros
top5 = top5.groupby(['Estabelecimento','T. Categoria']).head(5)

#ordenar por index permitindo uma leitura mais facil dos dados
top5 = top5.sort_index()
top5
```
